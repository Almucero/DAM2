```python
usuario1 = ['The Last of Us', 'One Piece', 'Shogun']
usuario2 = ['Fallout', 'The Boys', 'One Piece']

puntuacion_por_serie = {
    'The Last of Us': 95,
    'One Piece': 88,
    'Shogun': 92,
    'Fallout': 87,
    'The Boys': 90,
    'House of the Dragon': 89,
    'Severance': 93
}

def puntuacionLista(usuario, puntuacion, num):
    puntuacionTotal = 0
    for elemento in usuario:
      for key, item in puntuacion.items():
        if key == elemento:
          puntuacionTotal += item
    print(f"El usuario {num} tiene una puntuación total de {puntuacionTotal}")

try:
  puntuacionLista(usuario1, puntuacion_por_serie, 1)
  puntuacionLista(usuario2, puntuacion_por_serie, 2)
except:
  print("Ocurrió un problema")
```

```python
def dibujarCuadrados(altura, numero):
  try:
    if (altura<2):
      print("La altua no puede ser inferior a 2")
    elif (numero<1):
      print("Debe haber al menos un cuadrado")
    elif (altura*numero>35):
      print("El producto de la altura por el número de cuadrados no puede ser mayor a 35")
    else:
      for i in range(altura):
        if (i==(altura-1) or i==0):
          print("* "*int(altura))
        else:
          print("* ", " "*int(altura-2), "  *")
      print("")
  except:
    print("Ocurrió un error")

print("Dibuja cuadrados")
altura = int(input("Altura de los cuadrados: "))
numero = int(input("Número de cuadrados: "))

dibujarCuadrados(altura, numero)
```
